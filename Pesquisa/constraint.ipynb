```python
# Função para avaliar a solução
def avaliar_solucao(solution, num_homens, num_mulheres, preferencias_homens, preferencias_mulheres):
    # Validade das restrições
    for i in range(num_homens):
        if sum(solution[f'x_{i}_{j}'] for j in range(num_mulheres)) != 1:
            return False
    for j in range(num_mulheres):
        if sum(solution[f'x_{i}_{j}'] for i in range(num_homens)) != 1:
            return False

    # Verificação de estabilidade
    for i in range(num_homens):
        for j in range(num_mulheres):
            if solution[f'x_{i}_{j}'] == 1:
                parceiro_atual_homem = j
                parceiro_atual_mulher = i
                for k in range(num_mulheres):
                    if k != parceiro_atual_homem and preferencias_homens[i].index(k) < preferencias_homens[i].index(parceiro_atual_homem):
                        parceiro_alternativo_mulher = [solution[f'x_{m}_{k}'] for m in range(num_homens)].index(1)
                        if preferencias_mulheres[k].index(i) < preferencias_mulheres[k].index(parceiro_alternativo_mulher):
                            return False
                for l in range(num_homens):
                    if l != parceiro_atual_mulher and preferencias_mulheres[j].index(l) < preferencias_mulheres[j].index(parceiro_atual_mulher):
                        parceiro_alternativo_homem = [solution[f'x_{l}_{w}'] for w in range(num_mulheres)].index(1)
                        if preferencias_homens[l].index(j) < preferencias_homens[l].index(parceiro_alternativo_homem):
                            return False
    return True

# Exemplo de participantes (ex: homens e mulheres)
num_homens = 3
num_mulheres = 3

# Definindo preferências (exemplo simplificado)
preferencias_homens = [
    [0, 1, 2],  # Homem 0 prefere Mulher 0, depois Mulher 1, depois Mulher 2
    [1, 2, 0],  # Homem 1 prefere Mulher 1, depois Mulher 2, depois Mulher 0
    [2, 0, 1]   # Homem 2 prefere Mulher 2, depois Mulher 0, depois Mulher 1
]
preferencias_mulheres = [
    [0, 1, 2],  # Mulher 0 prefere Homem 0, depois Homem 1, depois Homem 2
    [1, 2, 0],  # Mulher 1 prefere Homem 1, depois Homem 2, depois Homem 0
    [2, 0, 1]   # Mulher 2 prefere Homem 2, depois Homem 0, depois Homem 1
]

# Criar o problema
problem = Problem()

# Adicionar variáveis
for i in range(num_homens):
    for j in range(num_mulheres):
        problem.addVariable(f'x_{i}_{j}', [0, 1])

# Adicionar restrições
# Cada homem deve ser emparelhado com uma mulher
for i in range(num_homens):
    problem.addConstraint(
        lambda *vals: sum(vals) == 1,
        [f'x_{i}_{j}' for j in range(num_mulheres)]
    )

# Cada mulher deve ser emparelhada com um homem
for j in range(num_mulheres):
    problem.addConstraint(
        lambda *vals: sum(vals) == 1,
        [f'x_{i}_{j}' for i in range(num_homens)]
    )

# Adicionar restrições baseadas nas preferências e empates
# Exemplo: Homem 0 prefere Mulher 0 e Mulher 1 igualmente (empate)
problem.addConstraint(
    lambda x_0_0, x_0_1: x_0_0 + x_0_1 <= 1,
    ('x_0_0', 'x_0_1')
)

# Resolver o problema
solutions = problem.getSolutions()
if solutions:
    for solution in solutions:
        if avaliar_solucao(solution, num_homens, num_mulheres, preferencias_homens, preferencias_mulheres):
            print('Solução válida e estável encontrada:')
            for i in range(num_homens):
                for j in range(num_mulheres):
                    if solution[f'x_{i}_{j}'] == 1:
                        print(f'Homem {i} emparelhado com Mulher {j}')
        else:
            print('Solução encontrada, mas não é válida ou estável.')
else:
    print('Nenhuma solução viável encontrada.')
```

```python
from constraint import Problem, AllDifferentConstraint
import timeit

start = timeit.default_timer()
problem = Problem()

men = ['x1', 'x2', 'x3']
women = ['y1', 'y2', 'y3']

#dominio das prioridades de preferência de cada um dos três homens
men_domain = {
    'x1' : ['y1','y2','y3'],
    'x2' : ['y1', 'y2'],
    'x3' : ['y1', 'y3']
}

for man in men_domain:
    problem.addVariable(man, men_domain[man])

problem.addConstraint(AllDifferentConstraint(), men_domain)

# função que verifica se um emparelhamento é estável
def stable_matching(x1,x2,x3):

    preferences = {
        'x1': [['y1'],['y2','y3']],
        'x2': [['y2'],['y1']],
        'x3': [['y1'],['y3']],
        'y1': [['y1'],['y2','y3']],
        'y2': [['y2'],['y1']],
        'y3': [['y1'],['y3']]
    }

    match = {'x1': x1, 'x2':x2, 'x3':x3}

    def prefers(prefs, p1, p2):
        for group in prefs:
            if p1 in group:
                return True
            if p2 in group:
                return False
        return False

    for man in men:
        woman = match[man]
        for other_woman in women:
            if other_woman != woman and other_woman in men_domain[man]:
                other_man = None
                for m, w in match.items():
                    if w == other_woman:
                        other_man = m
                        break
                if prefers( preferences[man], other_woman, woman) and prefers(preferences[other_woman], man, other_man):
                    return False
    return True

#constraint de estabilidade para verificar a estabilidade do emparelhamento
problem.addConstraint(stable_matching, men)

solutions= problem.getSolutions()

for solution in solutions:
    print(solution)
stop = timeit.default_timer()
print('Time: ', stop - start) 
```
