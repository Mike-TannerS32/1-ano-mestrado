```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.initializers import HeNormal
from tensorflow.keras.activations import elu
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.models import Sequential
```

```python
cifar= keras.datasets.cifar10
(X_train, y_train), (X_test, y_test) = cifar.load_data()
```

```python
print("Training data shape:", X_train.shape)
print("Training labels shape:", y_train.shape)
print("Testing data shape:", X_test.shape)
print("Testing labels shape:", y_test.shape)
```

```python
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0
```

```python
y_train = tf.keras.utils.to_categorical(y_train, 10)
y_test = tf.keras.utils.to_categorical(y_test, 10)
```

```python
initializer = HeNormal()
activation = elu
```

```python
model = Sequential()
model.add(Flatten(input_shape=(32, 32, 3)))
```

```python
for _ in range(20):
    model.add(Dense(100, activation=activation, kernel_initializer=initializer))

# Adicionando a camada de saída
model.add(Dense(10, activation='softmax'))

# Compilando o modelo
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Resumo do modelo
model.summary()
```

```python
history = model.fit(X_train, y_train, epochs=10, batch_size=64, validation_data=(X_test, y_test))
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=2)
print(f'\nTest accuracy: {test_acc}')
```

```python
import matplotlib.pyplot as plt

# Plot da precisão de treino e validação
plt.plot(history.history['accuracy'], label='Precisão de treino')
plt.plot(history.history['val_accuracy'], label='Precisão de validação')
plt.xlabel('Época')
plt.ylabel('Acurácia')
plt.legend(loc='lower right')
plt.title('Acurácia de Treino e Validação')
plt.show()

# Plotando a perda de treino e validação
plt.plot(history.history['loss'], label='Perda de treino')
plt.plot(history.history['val_loss'], label='Perda de validação')
plt.xlabel('Época')
plt.ylabel('Perda')
plt.legend(loc='upper right')
plt.title('Perda de Treino e Validação')
plt.show()
```

```python
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Nadam
```

```python
optimizer = Nadam(learning_rate=0.001)
model.compile(optimizer=optimizer, loss='categorical_crossentropy',metrics=['accuracy'])
model.summary()
```

```python
earlystopping= EarlyStopping(monitor='val_loss',patience='10',restore_best_weights=True)
```

```python
history = model.fit(X_train, y_train, epochs= 100, batch_size=64, validation_data=(X_test,y_test),callbacks=[earlystopping])
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=2)
print(f'\nTest accuracy: {test_acc}')
```

```python
plt.plot(history.history['accuracy'], label='Precisão de treino')
plt.plot(history.history['val_accuracy'], label='Precisão de validação')
plt.xlabel('Época')
plt.ylabel('Precisão')
plt.legend(loc='lower right')
plt.title('Acurácia de Treino e Validação')
plt.show()


plt.plot(history.history['loss'], label='Perda de treino')
plt.plot(history.history['val_loss'], label='Perda de validação')
plt.xlabel('Época')
plt.ylabel('Perda')
plt.legend(loc='upper right')
plt.title('Perda de Treino e Validação')
plt.show()
```

```python
from tensorflow.keras.optimizers.schedules import ExponentialDecay

# Configurar uma taxa de aprendizagem decrescente
initial_learning_rate = 0.001
lr_schedule = ExponentialDecay(
    initial_learning_rate,
    decay_steps=100000,
    decay_rate=0.96,
    staircase=True)

optimizer = Nadam(learning_rate=lr_schedule)
earlystopping= EarlyStopping(monitor='val_loss',patience='10',restore_best_weights=True)
model.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(X_train, y_train, epochs=100, batch_size=64, validation_data=(X_test, y_test), callbacks=[earlystopping])
```

```python
from tensorflow.keras.layers import BatchNormalization

# Construir o modelo com Batch Normalization
model_bn = Sequential()

# Adicionar a camada de entrada (flattening das imagens 32x32x3)
model_bn.add(Flatten(input_shape=(32, 32, 3)))

# Adicionar 20 camadas ocultas de 100 neurónios cada com Batch Normalization
for _ in range(20):
    model_bn.add(Dense(100, activation=activation, kernel_initializer=initializer))
    model_bn.add(BatchNormalization())

# Adicionar a camada de saída
model_bn.add(Dense(10, activation='softmax'))

# Compilar o modelo com o otimizador Nadam
optimizer_bn = Nadam(learning_rate=0.001)
earlystopping= EarlyStopping(monitor='val_loss',patience='10',restore_best_weights=True)
model_bn.compile(optimizer=optimizer_bn, loss='categorical_crossentropy', metrics=['accuracy'])

# Resumo do modelo
model_bn.summary()
```

```python
history_bn = model_bn.fit(X_train, y_train, epochs=100, batch_size=64, validation_data=(X_test, y_test), callbacks=[earlystopping])
```

```python
test_loss_bn, test_acc_bn = model_bn.evaluate(X_test, y_test, verbose=2)
print(f'\nTest accuracy with Batch Normalization: {test_acc_bn}')
```

```python
import matplotlib.pyplot as plt

# Comparar as curvas de aprendizagem

# Plotar a acurácia de treino e validação para ambos os modelos
plt.plot(history.history['accuracy'], label='Acurácia de treino (sem BN)')
plt.plot(history.history['val_accuracy'], label='Acurácia de validação (sem BN)')
plt.plot(history_bn.history['accuracy'], label='Acurácia de treino (com BN)')
plt.plot(history_bn.history['val_accuracy'], label='Acurácia de validação (com BN)')
plt.xlabel('Época')
plt.ylabel('Acurácia')
plt.legend(loc='lower right')
plt.title('Comparação de Acurácia de Treino e Validação')
plt.show()

# Plotar a perda de treino e validação para ambos os modelos
plt.plot(history.history['loss'], label='Perda de treino (sem BN)')
plt.plot(history.history['val_loss'], label='Perda de validação (sem BN)')
plt.plot(history_bn.history['loss'], label='Perda de treino (com BN)')
plt.plot(history_bn.history['val_loss'], label='Perda de validação (com BN)')
plt.xlabel('Época')
plt.ylabel('Perda')
plt.legend(loc='upper right')
plt.title('Comparação de Perda de Treino e Validação')
plt.show()
```
