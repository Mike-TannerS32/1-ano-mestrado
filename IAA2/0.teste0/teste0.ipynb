```python
df2 = pd.read_csv("dataset.zip")
df2.head()
```

```python
df2 = df2.rename(columns= {'W': 'Weight'})
df2.head()
```

```python
df2=df2.fillna({"BMI": df2['BMI'].mean()})
df2['BMI'].isnull().sum()
```

```python
df2=df2.fillna({"Weight": 60})
df2['Weight'].isnull().sum()
```

```python
df2['Stroke'] = df2['Stroke'].replace({'Negative':0, 'Positive':1})
df2['Stroke'].isnull().sum()
df2['Stroke'].unique()
```

```python
df2['Stroke'] = df2['Stroke'].replace({'#VALUE!':1})
df2['Stroke'].unique()
```

```python
df2.dropna(subset=["Height"], inplace=True)
df2.shape
```

```python
df2.to_csv(f'processed_dataset.zip', compression=dict(method='zip', archive_name=f'processed_dataset.csv'), encoding='utf-8', index=False, header=True)
```
