```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("long_clusters.csv")
df.head()
```

As colunas que podemos usar para fazer cluster dos dados com um modelo KMeans são Feature_1, Feature_2  e Feature_3.

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
features = df[['Feature_1','Feature_2', 'Feature_3']]
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(features['Feature_1'], features['Feature_2'], features['Feature_3'])
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.set_zlabel('Feature 3')

plt.show()
```

```python
import seaborn as sns

sns.pairplot(features)
plt.show()
```

```python
from sklearn.cluster import KMeans

scaler = StandardScaler()
normalized_features = scaler.fit_transform(features)

kmeans = KMeans(n_clusters=3)
kmeans.fit(normalized_features)
```
