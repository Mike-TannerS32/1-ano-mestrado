```python
df = pd.read_csv("indian_patient_records.zip")
df.head()
df.shape
df = df.dropna()
df.isnull().sum()
```

```python

from scipy import stats
import numpy as np
z = np.abs(stats.zscore(df['Age']))
print(z)
```

```python
threshold_z = 2
 
outlier_indices = np.where(z > threshold_z)[0]
no_outliers = df.drop(outlier_indices)
print("Original DataFrame Shape:", df.shape)
print("DataFrame Shape after Removing Outliers:", no_outliers.shape)
```

```python
count1 = 0
count2 = 0
count3 = 0
for i in y_test:
    if(i == "Iris-setosa"):
        count1 += 1
    elif(i =="Iris-versicolor"):
        count2 += 1
    elif(i == "Iris-virginica"):
        count3 += 1
total = count1 + count2 + count3
setosa_p = count1 / total
versicolor_p = count2 / total
virginica_p = count3 / total
print("Setosa: ",setosa_p*100, "%\n", "Versicolor: ",versicolor_p *100, "%\n", "Virginica:",virginica_p*100, "%\n")
```

```python
from sklearn.model_selection import KFold
X = df.drop("species", axis=1)
y = df.species

kfold = KFold(n_splits=3,random_state=0, shuffle=True)
```

```python
for fold, (train_indices, test_indices) in enumerate(kfold.split(X, y), 1):
        X_train, X_test = X.iloc[train_indices], X.iloc[test_indices]
        y_train, y_test = y.iloc[train_indices], y.iloc[test_indices]
        print("====Fold ",fold,"=====")
        print("=Train=")
        print(y_train.value_counts())
        print("=Test=")
        print(y_test.value_counts())
```

```python
from sklearn.model_selection import StratifiedKFold

SKFold = StratifiedKFold(n_splits=3,random_state=0, shuffle=True)
for fold, (train_indices, test_indices) in enumerate(SKFold.split(X, y), 1):
        X_train, X_test = X.iloc[train_indices], X.iloc[test_indices]
        y_train, y_test = y.iloc[train_indices], y.iloc[test_indices]
        print("====Fold ",fold,"=====")
        print("=Train=")
        print(y_train.value_counts())
        print("=Test=")
        print(y_test.value_counts())
```
